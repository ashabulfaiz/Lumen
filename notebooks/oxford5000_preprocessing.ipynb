###Load Dataset

```python
file_path = "/content/drive/MyDrive/Lumen/datasets/Oxford5000.txt"

with open(file_path, "r", encoding="utf-8") as f:
    words = f.read().splitlines()

print(len(words))
```

###Cek Data

```python
print(words[:20])
```

###Cleaning Data

```python
words = [w.strip() for w in words if w.strip()]
```

###Hapus Duplikat

```python
words = list(set(words))
```

###Sort Data

```python
words = sorted(words)
```

###Dataframe

```python
import pandas as pd

df = pd.DataFrame(words, columns=["word"])
df.head()
```

```python
filtered_words = []

bad_prefix = [
    "abo",
    "abs",
    "acad",
    "accel",
    "accept",
    "access",
    "accom",
    "admin",
    "adv",
    "aff",
    "algo",
    "alter",
    "analyse",
    "announ",
    "anti",
    "apolog",
    "approach",
    "argu",
    "assess",
    "assume",
    "authority"
]

for word in words:

    # skip phrase
    if " " in word:
        continue

    # skip symbol
    if "-" in word:
        continue

    if len(word) > 10:
        continue

    # skip angka
    if any(char.isdigit() for char in word):
        continue

    # skip kata akademik
    if any(word.startswith(p) for p in bad_prefix):
        continue

    filtered_words.append(word)
```

```python
filtered_words = filtered_words[:1500]

print("Jumlah filtered words:", len(filtered_words))
```

```python
filtered_words[:50]
```

```python
import pandas as pd

df = pd.DataFrame(
    filtered_words,
    columns=["word"]
)
```

###Translate Data

```python
translator = GoogleTranslator(
    source='en',
    target='id'
)

df["translation"] = df["word"].apply(
    lambda x: translator.translate(x)
)
```

```python
df.head()
```

###Level



```python
df["level"] = "beginner"
```

###Class Learning

```python
classes = [
    "Daily Vocabulary",
    "Basic Sentences",
    "Daily Conversation",
    "Basic Grammar"
]

df["class"] = ""

total = len(df)

split1 = total // 4
split2 = split1 * 2
split3 = split1 * 3

for i in range(total):

    if i < split1:
        df.iloc[i, df.columns.get_loc("class")] = classes[0]

    elif i < split2:
        df.iloc[i, df.columns.get_loc("class")] = classes[1]

    elif i < split3:
        df.iloc[i, df.columns.get_loc("class")] = classes[2]

    else:
        df.iloc[i, df.columns.get_loc("class")] = classes[3]
```

###Example Sentence

```python
def generate_example(word, class_name):

    if class_name == "Daily Vocabulary":
        return f"I learn the word '{word}'."

    elif class_name == "Basic Sentences":
        return f"I use '{word}' in a simple sentence."

    elif class_name == "Daily Conversation":
        return f"I say '{word}' in daily conversation."

    else:
        return f"I study the grammar of '{word}'."
```

```python
df["example"] = df.apply(
    lambda row: generate_example(
        row["word"],
        row["class"]
    ),
    axis=1
)
```

```python
df = df[
    [
        "word",
        "translation",
        "level",
        "class",
        "example"
    ]
]
```
